## Two core objects in pandas

### Series
- **Series**: A 1D labeled array (like a column)
- It is a sequence of data values. If a DataFrame is a table, a Series is a list. 


### DataFrame
- **DataFrame**: A 2D labeled data structure (like a full spreadsheet or SQL table)
- It is a data table. It contains an array of individual entries, each of which has a certain value. 
Each entry corresponds to a row (or record) and a column.

# Series




```python

my_fruits = ["Apple", "Kiwi", "Tomato"]

fruit_series = pd.Series(my_fruits) # .Series is a constructor not a function

print(fruit_series, "\n")

fruit_series = pd.Series (my_fruits, index=["Pome Type", "Berry Type 1", "Berry Type 2"]) # Or fruit_series.index = [...]

fruit_series.head() # Show top 5 rows
```

```python
my_fruits: dict[str, str] = {
    "Fruit 1": "Apple",
    "Fruit 2": "Mango",
    "Fruit 3": "Orange"
} 

fruit_series = pd.Series(my_fruits) # We don't need indices since we input dictonary, keys will be taken as labels

print(fruit_series,"\n")

print("Location By label: ", fruit_series.loc["Fruit 1"])
print("Location By Index: ", fruit_series.iloc[1], "\n")

# Mutating fruit_series using loc and iloc

fruit_series.loc["Fruit 1"] = "Banana"
fruit_series.iloc[-1] = "Guauva"

print(f"Series: {fruit_series} having dtype {fruit_series.dtype} of dimensions numbers: {fruit_series.ndim} and dimensions: {fruit_series.shape}")


```

```python
### Series: Calories

calories: dict[str, int] = {
    "Day 1": 1300,
    "Day 2": 2000,
    "Day 3": 2322,
}

calories_series = pd.Series(calories)

print(calories_series, "\n")

# Vectorized Comparision 

print("Vectorized Comparision ..\n")
print("> 1200", calories_series > 1200, sep="\n")
print("== 1300",  calories_series == 1300, sep="\n")


# Indexing (Array Masking)

print("\nArray Masking..\n")
print("> 1200", calories_series[calories_series > 1200], sep="\n")
print("<=2000" , calories_series[calories_series <= 2000], sep="\n")
 
```

```python
series_fromlist = pd.Series([189, 232, 322], index=["S1", "S2","S3"]) 
series_fromdict =  pd.Series({"S1": 189, "S2": 232, "S3": 322})

print(series_fromlist, "\n")
print(series_fromdict)
```

## DataFrame

A tabular data structure with rows and columns. (2 Dimensions). Similar to Excel Spreadsheet

```python
data : dict [str, list[str | int]] = {
    "Name" : ["Spongebob", "Patrick", "Squidward"],
    "Age"  : [30, 35, 50]
} 

# data_series = pd.Series(next(iter(data.values())))
data_series = pd.Series(data["Name"])

print("Series:\n",data_series,"\n", sep="")


df = pd.DataFrame(data) 

# To override headers with a dict in DataFrame

# Use 'columns' to select and order the dictionry keys
# df = pd.DataFrame(data, columns=["Name", "Age"])

# To completely change the names, rename them right after creation
# df.columns = ["Age", "City"]


print("DataFrame (df)\n")
df.head()
print(f"Dtype: {df.dtypes}")
```

```python
## Panda DataFrame with Index
print(data)

df = pd.DataFrame(data, index=["Employee 1", "Employee 2", "Employee 3"] )

df.head()
```

```python
# Selecting row with DataFrame with iloc and loc

df_row_1 = df.iloc[0]

df_first_row_employee_1= df.loc["Employee 1"]


print("Row 0 using df.iloc[0] - Location by Index:\n")
print(df_row_1, "\n")

print("Row 1 using df.loc['Employee 2'] - Location by index label:\n")
print(df_first_row_employee_1, "\n")
```

```python
# Selecting col DataFrame with Column Label

df_columns = df[["Name", "Age"]]

print("Name Column using df['Name']:\n")
print(df["Name"], "\n")

# Selecting cols with DataFrame

print("Name Column and Age Column using df[['Name', 'Age']]: \n")
print(df_columns, "\n")

# In numpy selecing col 1
np_array = np.array([[1, 2, 3], [4, 5, 6]])
print("Numpy  2d Array, Row 0: ", np_array[:, 0], "\n")

# In pandas selecting col with iloc, since df[:, 0] is invalid syntax resulting invalid tuple 
print("Pandas row selection with .iloc[]:\n", df.iloc[:, 0])
```
