It makes working with structured data **fast, flexible, and intuitive** — especially if you're dealing with CSV files, Excel sheets, SQL tables, JSON, or APIs.

- Tools for **reading and writing data** between in-memory data structures and different formats: CSV and text files, Microsoft Excel, SQL databases, and the fast HDF5 format;

- Intelligent **data alignment** and integrated handling of missing data: gain automatic label-based alignment in computations and easily manipulate messy data into an orderly form;

- Widely used in **academic and commercial fields**, including Finance, Neuroscience, Economics, Statistics, Advertising, and Web Analytics.

## Two core objects in pandas

### Series
- **Series**: A 1D labeled array (like a column)
- It is a sequence of data values. If a DataFrame is a table, a Series is a list. 


### DataFrame
- **DataFrame**: A 2D labeled data structure (like a full spreadsheet or SQL table)
- It is a data table. It contains an array of individual entries, each of which has a certain value. 
Each entry corresponds to a row (or record) and a column.

# Series




```python

my_fruits = ["Apple", "Kiwi", "Tomato"]

fruit_series = pd.Series(my_fruits) # .Series is a constructor not a function

print(fruit_series, "\n")

fruit_series = pd.Series (my_fruits, index=["Pome Type", "Berry Type 1", "Berry Type 2"]) # Or fruit_series.index = [...]

fruit_series.head() # Show top 5 rows
```

```python
my_fruits: dict[str, str] = {
    "Fruit 1": "Apple",
    "Fruit 2": "Mango",
    "Fruit 3": "Orange"
} 

fruit_series = pd.Series(my_fruits) # We don't need indices since we input dictonary, keys will be taken as labels

print(fruit_series,"\n")

print("Location By label: ", fruit_series.loc["Fruit 1"])
print("Location By Index: ", fruit_series.iloc[1], "\n")

# Mutating fruit_series using loc and iloc

fruit_series.loc["Fruit 1"] = "Banana"
fruit_series.iloc[-1] = "Guauva"

print(f"Series: {fruit_series} having dtype {fruit_series.dtype} of dimensions numbers: {fruit_series.ndim} and dimensions: {fruit_series.shape}")


```

```python
### Series: Calories

calories: dict[str, int] = {
    "Day 1": 1300,
    "Day 2": 2000,
    "Day 3": 2322,
}

calories_series = pd.Series(calories)

print(calories_series, "\n")

# Vectorized Comparision 

print("Vectorized Comparision ..\n")
print("> 1200", calories_series > 1200, sep="\n")
print("== 1300",  calories_series == 1300, sep="\n")


# Indexing (Array Masking)

print("\nArray Masking..\n")
print("> 1200", calories_series[calories_series > 1200], sep="\n")
print("<=2000" , calories_series[calories_series <= 2000], sep="\n")
 
```
