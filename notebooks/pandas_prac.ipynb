# Pandas - Panel Data (Econometrics term for multidimensional data sets) / Python Data Analysis

Pandas is a **fast, powerful, flexible and easy to use data analysis and manipulation tool**, built on top of the Python programming language and another library, NumPy.

In particular, it **offers data structures and operations for manipulating numerical tables and time series**. It's **widely used in data analysis, data science, and machine learning**.

- A fast and efficient **DataFrame** object for data manipulation with integrated indexing;

- Tools for **reading and writing data** between in-memory data structures and different formats: CSV and text files, Microsoft Excel, SQL databases, and the fast HDF5 format;

- Intelligent **data alignment** and integrated handling of missing data: gain automatic label-based alignment in computations and easily manipulate messy data into an orderly form;

- Widely used in **academic and commercial fields**, including Finance, Neuroscience, Economics, Statistics, Advertising, and Web Analytics.

## Highlights

1. Fast and efficient **DataFrame** with integrated indexing.
2. Read/write data in **CSV, text, Excel, SQL, and HDF5** formats.
3. Intelligent **data alignment** and **missing-data handling**.
4. Flexible **reshaping** and **pivoting** of datasets.
5. Powerful **slicing, fancy indexing, and subsetting**.
6. **Size mutability** — columns can be inserted or deleted.
7. Powerful **group by** engine for split-apply-combine operations.
8. High-performance **merging and joining** of datasets.
9. **Hierarchical indexing** for handling high-dimensional data.
10. **Time-series tools**: date ranges, frequency conversion, moving statistics, shifting, and lagging.
11. Highly **optimized for performance** using Cython and C.
12. Widely used in **academic and commercial fields**, including Finance, Neuroscience, Economics, Statistics, Advertising, and Web Analytics.
